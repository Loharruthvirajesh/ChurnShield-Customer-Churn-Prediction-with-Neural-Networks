```python
categorical_cols = X.select_dtypes(include='object').columns.tolist()

numeric_cols = X.select_dtypes(include=np.number).columns.tolist()

print("Categorical columns:", categorical_cols)
print("Numerical columns:", numeric_cols)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print(y_train.value_counts(normalize=True))
print(y_test.value_counts(normalize=True))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
    ]
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('classifier', LogisticRegression(max_iter=1000))
    ]
)
```

# 9. Regression Module — Predicting Employee Evaluation

```python
X_reg = df_clean.drop('last_evaluation', axis=1)
y_reg = df_clean['last_evaluation']
```

```python
categorical_reg = X_reg.select_dtypes(
    include='object'
).columns.tolist()

numeric_reg = X_reg.select_dtypes(
    include=np.number
).columns.tolist()

print("Categorical:", categorical_reg)
print("Numerical:", numeric_reg)
```

```python
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.20,
    random_state=42
)
```

```python
reg_preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_reg),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_reg)
    ]
)
```

```python
from sklearn.linear_model import LinearRegression

linear_model = Pipeline(
    steps=[
        ('preprocessor', reg_preprocessor),
        ('regressor', LinearRegression())
    ]
)
```

```python
linear_model.fit(
    X_train_reg,
    y_train_reg
)
```

```python
linear_pred = linear_model.predict(X_test_reg)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
mae = mean_absolute_error(
    y_test_reg,
    linear_pred
)

mse = mean_squared_error(
    y_test_reg,
    linear_pred
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y_test_reg,
    linear_pred
)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R²:", r2)
```

```python
plt.figure(figsize=(7, 6))

plt.scatter(
    y_test_reg,
    linear_pred
)

plt.xlabel('Actual Evaluation')
plt.ylabel('Predicted Evaluation')
plt.title('Actual vs Predicted Evaluation')

plt.show()
```

```python
ridge_model = Pipeline(
    steps=[
        ('preprocessor', reg_preprocessor),
        ('regressor', Ridge(alpha=1.0))
    ]
)
```

```python
ridge_model.fit(
    X_train_reg,
    y_train_reg
)
```

```python
ridge_pred = ridge_model.predict(X_test_reg)
```

```python
ridge_mae = mean_absolute_error(
    y_test_reg,
    ridge_pred
)

ridge_rmse = np.sqrt(
    mean_squared_error(
        y_test_reg,
        ridge_pred
    )
)

ridge_r2 = r2_score(
    y_test_reg,
    ridge_pred
)

print("Ridge MAE:", ridge_mae)
print("Ridge RMSE:", ridge_rmse)
print("Ridge R²:", ridge_r2)
```

```python
from sklearn.linear_model import Lasso
```

```python
lasso_model = Pipeline(
    steps=[
        ('preprocessor', reg_preprocessor),
        ('regressor', Lasso(alpha=0.01))
    ]
)
```

```python
lasso_model.fit(
    X_train_reg,
    y_train_reg
)
```

```python
lasso_pred = lasso_model.predict(X_test_reg)
```

```python
lasso_mae = mean_absolute_error(
    y_test_reg,
    lasso_pred
)

lasso_rmse = np.sqrt(
    mean_squared_error(
        y_test_reg,
        lasso_pred
    )
)

lasso_r2 = r2_score(
    y_test_reg,
    lasso_pred
)

print("Lasso MAE:", lasso_mae)
print("Lasso RMSE:", lasso_rmse)
print("Lasso R²:", lasso_r2)
```

### Ridge vs Lasso

Ridge regression applies L2 regularization, which penalizes large coefficients and helps reduce model complexity.

Lasso regression applies L1 regularization. In addition to shrinking coefficients, Lasso can reduce some coefficients to zero, which can perform a form of feature selection.

Both techniques are used to reduce overfitting and improve generalization.

# 10. Multiclass Classification — Salary Prediction

```python
X_multi = df_clean.drop('salary', axis=1)
y_multi = df_clean['salary']
```

```python
y_multi.value_counts()
```

```python
categorical_multi = X_multi.select_dtypes(
    include='object'
).columns.tolist()

numeric_multi = X_multi.select_dtypes(
    include=np.number
).columns.tolist()

print("Categorical:", categorical_multi)
print("Numerical:", numeric_multi)
```

```python
X_train_multi, X_test_multi, y_train_multi, y_test_multi = train_test_split(
    X_multi,
    y_multi,
    test_size=0.20,
    random_state=42,
    stratify=y_multi
)
```

```python
multi_preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_multi),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_multi)
    ]
)
```

```python
multi_log_model = Pipeline(
    steps=[
        ('preprocessor', multi_preprocessor),
        ('classifier', LogisticRegression(
            max_iter=1000
        ))
    ]
)
```

```python
multi_log_model.fit(
    X_train_multi,
    y_train_multi
)
```

```python
multi_pred = multi_log_model.predict(
    X_test_multi
)
```

```python
print(
    classification_report(
        y_test_multi,
        multi_pred
    )
)
```

```python
multi_cm = confusion_matrix(
    y_test_multi,
    multi_pred
)

multi_cm
```

```python
sns.heatmap(
    multi_cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=multi_log_model.classes_,
    yticklabels=multi_log_model.classes_
)

plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Salary Classification Confusion Matrix')

plt.show()
```

```python
print(
    classification_report(
        y_test_multi,
        multi_pred,
        output_dict=True
    )['macro avg']
)
```
